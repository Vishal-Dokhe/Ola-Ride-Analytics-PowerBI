```python
## 1. Import Libraries
import pandas as pd
import numpy as np
```

```python
# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

```python
## 2. Load Dataset
df = pd.read_csv(r"C:\Users\Vishal Dokhe\Downloads\Bengaluru Ola.csv")
print("Dataset Loaded Successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
df
```

```python
# Check Dataset Information
df.info()
```

```python
# Convert Date Column

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d/%m/%Y"
)
```

```python
# Convert Time Column

df["Time"] = pd.to_datetime(
    df["Time"],
    format="%H:%M:%S"
)
```

```python
df.dtypes
```

```python
# Remove Leading & Trailing Spaces from Text Columns

# Select all object (text) columns
text_columns = df.select_dtypes(include="object").columns

# Remove leading and trailing spaces
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
# Check for Negative Values

numeric_columns = [
    "Avg VTAT",
    "Avg CTAT",
    "Booking Value",
    "Ride Distance",
    "Driver Ratings",
    "Customer Rating"
]

for col in numeric_columns:
    if (df[col].dropna() < 0).any():
        print(f"❌ Negative values found in {col}")
    else:
        print(f"✅ {col} - No negative values")
```

```python
# Validate Driver Ratings

print("Driver Rating Range")
print(df["Driver Ratings"].min(), "-", df["Driver Ratings"].max())
```

```python
# Validate Customer Ratings
print("\nCustomer Rating Range")
print(df["Customer Rating"].min(), "-", df["Customer Rating"].max())
```

```python
# Validate Ride Distance
print("Ride Distance Range")
print(df["Ride Distance"].min(), "-", df["Ride Distance"].max())
```

```python
# Validate Booking Value
print("Booking Value Range")
print(df["Booking Value"].min(), "-", df["Booking Value"].max())
```

```python
# Export Cleaned Dataset

df.to_csv("ola_cleaned.csv", index=False)

print("✅ ola_cleaned.csv has been created successfully!")
```
