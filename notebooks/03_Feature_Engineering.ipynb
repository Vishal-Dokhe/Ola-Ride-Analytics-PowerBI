```python
import pandas as pd
```

```python
# Load cleaned dataset
df = pd.read_csv("ola_cleaned.csv")
```

```python
df
```

```python
# Convert Date and Time back to datetime
df["Date"] = pd.to_datetime(df["Date"])
df["Time"] = pd.to_datetime(df["Time"])
```

```python
# Check dataset
df.head()
```

```python
# Year
df["Year"] = df["Date"].dt.year
```

```python
# Quarter
df["Quarter"] = df["Date"].dt.quarter
```

```python
# Month Number
df["Month"] = df["Date"].dt.month
```

```python
# Month Name
df["Month Name"] = df["Date"].dt.month_name()
```

```python
# Day of Month
df["Day"] = df["Date"].dt.day
```

```python
# Day Name
df["Day Name"] = df["Date"].dt.day_name()
```

```python
# Hour
df["Hour"] = df["Time"].dt.hour
```

```python
def time_slot(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["Time Slot"] = df["Hour"].apply(time_slot)
```

```python
print(df[[
    "Year",
    "Quarter",
    "Month",
    "Month Name",
    "Day",
    "Day Name",
    "Hour",
    "Time Slot"
]].head())
```

```python
print("Years:", df["Year"].unique())
print("Quarters:", df["Quarter"].unique())
print("Months:", df["Month Name"].unique())
print("Days:", df["Day Name"].unique())
print("Time Slots:", df["Time Slot"].unique())
```

```python
df.to_csv("ola_feature_engineered.csv", index=False)

print("Feature Engineering Completed Successfully!")
```
